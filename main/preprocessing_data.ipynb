```python
import pandas as pd
import requests
```

```python
BASE_URL = 'https://rickandmortyapi.com/api/'
N = 826 #All characters
print('Done')
```

```python
#API Request - get one character

req = requests.get(f'{BASE_URL}character/50', timeout=30)
req.raise_for_status()
data = req.json()

print('Name:', data['name'])
print('Status:', data['status'])
print('Species:', data['species'])
print('Type:', data['type'])
print('Gender:', data['gender'])

def flatten(data):
    types = data['type']
    status = data['status']

    return {
        'name': data['name'],
        'status': status,
        'species': data['species'],
        'type': types,
        'gender': data['gender']
    }

print()
```

```python
#API Request - get all characters and save in dataset.csv

all_characters = list(range(1, N+1))

chunk_size = 50
rows = []

for i in range (0, len(all_characters), chunk_size):
    chunk = all_characters[i : i + chunk_size]
    chunk_str = ",".join(map(str, chunk))

    req = requests.get(f"{BASE_URL}/character/{chunk_str}", timeout=30)
    req.raise_for_status()

    characters_data = req.json()
    for character in characters_data:
        rows.append(flatten(character))
    
    print(f"Fetched {len(rows)}/{N} characters...", end="\r")

df = pd.DataFrame(rows)

print('\nForm:', df.shape)

df.head()
```

```python
df.to_csv('../data/dataset.csv', index=False)
print('Done')
```
